## Wine
Is a compatibility layer for Linux (and macOS) that allows one to run Windows applications. It works by translating Windows API calls into POSIX calls on the fly.

## Wine Prefix
A prefix is a self contained simulated Windows installation. It contains a `C:` drive, registry and other things. An example prefix representing Irfan View application installation:
```sh
$ tree ~/Apps/Wine\ Apps/irfan_view/
/home/salman/Apps/Wine Apps/irfan_view/
├── dosdevices
│   ├── c: -> ../drive_c
|   ...
│   └── z: -> /
├── drive_c
│   ├── ProgramData
│   │   └── ...
│   ├── Program Files
│   │   └── ...
│   ├── Program Files (x86)
│   │   └── ...
│   ├── users
│   │   └── ...
│   └── windows
│       └── ...
├── system.reg
├── userdef.reg
└── user.reg

```

We can install multiple programs in the same prefix, but usually we stick to one prefix per program.

By default, the prefix location is `~/.wine`. But we can set environment variable `WINEPREFIX=~/prefixes/myapp` and it will change it to the stated directory. Subsequent Wine commands would consider that prefix.

## Wine Commands
To run an exe:
```sh
# This will install in the WINEPREFIX if set, else ~/.wine
$ wine ~/Downloads/my_installer.exe
```

Wine also create `.desktop` entries for the program automatically.

To configure the Windows environment created in the prefix, we can use the `winecfg` command. It opens up a dialog box containing multiple setting that can be updated:
![winecfg](./images/winecfg.png)

**Applications:** wets the Windows version Wine reports to a program. Can be done per exe basis.

**Libraries:** controls DLL overrides. Available options:
- *builtin*: use Wine's re-implementation
- *native*: use a real Microsoft DLL (or one you supplied)

**Graphics:** has options to:
- automatically capture the mouse in fullscreen windows
- allow the window manager to decorate and control windows
- emulate a virtual desktop (runs the app inside a fixed-size window, handy for games that mess with your resolution or multi-monitor setups)
- screen resolution (DPI) for font and UI scaling.

**Desktop Integration:** contins appearance/theme for Windows widgets and shell folders. Where "My Documents", "Desktop", "Downloads", and so on point. By default they link to Linux home directories. Unchecking the link keeps them inside the prefix, which is a bit safer for isolation.

**Drives:** maps Windows drive letters to Linux directories. `C:` points to the prefix's `drive_c`, `Z:` usually points to `/`, etc.

**Audio:** selects the audio driver and test playback.
