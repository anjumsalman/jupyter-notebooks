## ConfigMap Configuration
Basic ConfigMap setup"
```yaml
apiVersion: v1
kind: ConfigMap
metadata:
  name: order-api-config
data:
  LOG_LEVEL: "info"
  MAX_RETRIES: "3"
  API_TIMEOUT_SECONDS: "30"
```

Then use it in Pod or Deployment:
```yaml
containers:
  - name: order-api
    image: my-registry/order-api:latest
    envFrom:
      - configMapRef:
          name: order-api-config
```

Along with `configMapRef.name` we can also add `configMapRef.key` to only import certain keys.

ConfigMap can also contain entire configuration file:
```yaml
apiVersion: v1
kind: ConfigMap
metadata:
  name: app-config
  namespace: default
data:
  application.properties: |
    server.port=8080
    spring.datasource.hikari.maximum-pool-size=20
    management.endpoints.web.exposure.include=health,info,metrics

  nginx.conf: |
    events { worker_connections 1024; }
    http {
      server {
        listen 80;
        location / { return 200 "OK"; }
      }
    }
```

These configuration files can be mounted inside a container:
```yaml
    volumeMounts:
      - name: config-volume
        mountPath: /etc/config
volumes:
  - name: config-volume
    configMap:
      name: app-config
```
