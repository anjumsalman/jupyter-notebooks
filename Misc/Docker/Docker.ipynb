## Hypervisor Virtualization
Provides the ability to run multiple operating systems simultaneously on a single physical hardware. This involves creation of *virtual machines* by abstracting away the hardware and divides it among the VMs. Operating system is installed on individual VMs each thinking it has its own hardware.

There are two types of hypervisors:
- **bare metal:** hypervisor runs directly on top of the hardware, no OS underneath. Examples include Microsoft Hyper-V, Amazon Nitro, Xen, etc.
- **hosted:** hypervisor runs on top of an OS like any other program. Examples include VirtualBox, VMWare WorkStation, etc.

## Container Virtualization
Instead of virtualizing hardware, container virtualization virtualizes the operating system. Container share the operating system kernel and just isolate the processes running on it. Applications, libraries and other dependencies are separated out into each container.

Instead of using hypervisors, container virtualization use facilities provided by the operating system. Taking the example of Linux, containerization is made possible using two features:
- **namespaces:** partition kernel resources such that one set of processes see one set of resources while another set of processes see a different set of resources. Each container gets its own process tree (process namespace), network (network namespace), filesystem (mount namespace), users, hostnames, etc.
- **cgroups:** provide resource limiting by sharing CPU, memory, network among all containers.

Other OS have their own technology:
- Solaris provides Zones
- BSD provides Jails
- Windows Containers

## Docker
Docker is an open-source engine that automates the deployment of applications into containers. It is composed of the following components:
- **Docker CLI:** command-line interafce into Docker, it sends commands (to Docker Daemon) to run various functions.
- **Docker Daemon:** `dockerd` is a background pricess that does all the work including managing containers, building images, managing volumes and networks. It receives commands from the client and delegates its execution downstream.
- **ContainerD:** `containerd` is another background process that sits lower in hierarchy and is responsible for managing container lifecycle like starting, stopping, pausing and monitoring containers. It also pulls images and manages network and volume namespaces at a lower level. Technologies like Kubernetes just make use of `containerd` and skip the Docker daemon component altogether.
- **`runc`:** is a lower-level CLI tool/library that implements the OCI runtime spec. It's the piece that actually talks to the Linux kernel to create a container; setting up `namespace`s, `cgroup`s, and finally executing the process inside that isolated environment. `containerd` spawns a `runc` process for each container.
- **Docker Registry:** is the place where images are stored and pulled from. Example DockerHub.
- **Docker Objects:** include
    - Images: the read-only templates
    - Containers: running instances of images
    - Volumes: persistent storage
    - Networks: virtual networks connecting containers
