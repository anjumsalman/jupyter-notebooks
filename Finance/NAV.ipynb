## Overview
It is the value of an investment fund determined after subtracting liabilities from total assets.

$$NAV = assets - liabilities$$
$$NAV\ per\ share = \frac{assets - liabilities}{total\ number\ of\ outstanding\ shares}$$

**Assets:** include total market value of all securities in the fund (stocks, bonds, cash, other investments), including accrued income (like dividends or interest receivable)  
**Liabilities:** include fund expenses, management fees payable, and any other obligations

Most funds calculate NAV once per day, typically after markets close, using closing prices of the underlying holdings.

As an example, consider a fund that invests in the following securities:

| Reliance Industries | 50,00,000   | 2850          | 1,42,50,00,00,000 |
| HDFC Bank           | 80,00,000   | 1650          | 1,32,00,00,00,000 |
| Infosys             | 60,00,000   | 1480          | 88,80,00,00,000   |
| TCS                 | 40,00,000   | 3900          | 1,56,00,00,00,000 |
| ICICI Bank          | 70,00,000   | 1120          | 78,40,00,00,000   |
| **Total**           |             |               | 5,97,70,00,00,000 |
</div>    

Other assets:
- Cash & Cash Equivalents: 2,30,00,00,000 (for redemptions, new investments)
- Accrued Dividends Receivable: 15,00,00,000

Total Assets = 5,97,70,00,00,000 + 2,30,00,00,000 + 15,00,00,000 = 6,00,15,00,00,000

Now considering the liabilities, we have:
- Management Fees Payable (accrued daily, based on 0.71% expense ratio annualized): 8,00,00,000
- Other Payables (audit fees, operational costs accrued): 2,00,00,000

Total liabilities = 8,00,00,000 + 2,00,00,000 = 10,00,00,000

Calculating NAV, we get
$$NAV = 6,00,15,00,00,000 - 10,00,00,000 = 5,99,05,00,00,000$$
Now assuming all investors combined have 4,87,50,00,000 units outstanding,
$$NAV\ per\ share = \frac{5,99,05,00,00,000}{4,87,50,00,000} = 122.88$$

### Investment Instruments Associated with NAV
- **Mutual Funds:** this is the most common association. NAV determines the price at which investors buy or redeem units. Unlike stocks, mutual fund NAV is set once daily (end-of-day), not continuously during trading.
- **Exchange-Traded Funds (ETFs):** ETFs have an official NAV calculated daily, but since they trade on exchanges throughout the day, their market price can differ slightly from NAV due to supply/demand (creating premiums or discounts). Authorized participants use arbitrage to keep the two aligned.
- **Hedge Funds:** NAV is used to value investor stakes, though it's typically calculated less frequently (monthly or quarterly) and involves more complex/illiquid asset valuations.

Other instruments like REITs, UITs, ULIPs also have NAV.
