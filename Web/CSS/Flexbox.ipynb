## Overview
Gives the ability to place containers in rows and columns and provides lot of control over their placement and alignment. The default `display` mode is `flow` which renders containers as:

```html
<div class="outer-container">
  <div class="top-container">Top Container</div>
  <div class="middle-container">Middle Container</div>
  <div class="bottom-container">Bottom Container</div>
</div>
```
```css
.outer-container {
  background-color: #FFFCD2;
  padding: 10px;
  color: white;
}

.top-container {
  background-color: #FF4838;
  padding: 10px;
}

.middle-container {
  background-color: #41ABFC;
  padding: 10px;
}

.bottom-container {
  background-color: #1FC65D;
  padding: 10px;
}
```
<div style="background-color: #FFFCD2; padding: 10px">
    <div style="background-color: #FF4838; color: white; padding: 10px">Top Container</div>
    <div style="background-color: #41ABFC; color: white; padding: 10px">Middle Container</div>
    <div style="background-color: #1FC65D; color: white; padding: 10px">Bottom Container</div>
</div>

Making a small change to `display` property changes this to:
```css
.outer-container {
  background-color: #FFFCD2;
  padding: 10px;
  color: white;
  display: flex
}
```
<div style="display: flex; background-color: #FFFCD2; padding: 10px">
    <div style="background-color: #FF4838; color: white; padding: 10px">Top Container</div>
    <div style="background-color: #41ABFC; color: white; padding: 10px">Middle Container</div>
    <div style="background-color: #1FC65D; color: white; padding: 10px">Bottom Container</div>
</div>

## Flex Direction
Under `flex` containers by default appear as distinct columns in the same row. We can change that to row as well by using `flex-direction` property:
![flex direction](./images/flex-direction.png)  

There are two axes to consider:
- **primary axis**: direction in which containers will be bunched up at the start of the parent container
- **cross axis**: direction in which containers will stretch out to fill the entire parent container

## Alignment
`justify-content` property decides how containers are distributed along the primary axis.
```css
.outer-container {
  background-color: #FFFCD2;
  padding: 10px;
  color: white;
  display: flex;
  justify-content: flex-start
}
```

What is the difference between `space-around` and `space-evenly`?
- space-around: items get equal space on each side of themselves, which means the middle gaps end up twice as wide as the edge gaps.
- space-evenly: all gaps are exactly equal, including the gaps at the start and end edges.

This means that if there are 100px of free space (after placing all containers), then:
- space-around has edge gaps = 	16.67px each (100 ÷ 6) and between gaps = 33.33px each (16.67 × 2)
- space-evenly has edge gaps = 25px each (100 ÷ 4) and between gaps = 25px each

<br>

`baseline` aligns using the line passing through the bottom of the first line of every container.

What if we want to change the alignment of a particular container along the cross axis? We can use the `align-self` property:
```css
.outer-container {
  background-color: #FFFCD2;
  padding: 10px;
  color: white;
  display: flex;  
}

.left-container {
  background-color: #FF4838;
  padding: 10px;
  align-self: center;
}
```
<div style="display: flex; align-items: stretch; background-color: #FFFCD2; padding: 10px; height: 100px">
    <div style="align-self: center; background-color: #FF4838; color: white; padding: 10px">Left Container</div>
    <div style="background-color: #41ABFC; color: white; padding: 10px">Middle Container<br>New Line</div>
    <div style="background-color: #1FC65D; color: white; padding: 10px">Right Container</div>
</div>
<br>

The values this property can take is same as `align-items`. There is no `justify-self` understandably since that would mean colliding containers. To summarise:
- justify: to position something along the primary axis.
- align: to position something along the cross axis.
- content: a group of items that can be distributed.
- items: single items that can be positioned individually.

<div style="display:inline-block">
    
|         | content | items   |
|---------|---------|---------|
| justify | ✅       | ❌     |
| align   | ✅       | ✅     |
</div>

## Container Width
In the default layout mode the `width` property of container is enforced which can result in the child container shooting out of the bounds of the parent container:

```css
.parent-container {
  background-color: #FFFCD2;
  padding: 10px;
  width: 500px;
  border: 1px solid orange;
}

.child-container {
  background-color: #FF4838;
  padding: 10px;
  width: 700px;
}
```
<div style="background-color: #FFFCD2; border: 1px solid orange; padding: 10px; width: 500px;">
    <div style="background-color: #FF4838; color: white; padding: 10px; width: 700px">Child Container</div>
</div>
<br>

However, in `flex` mode, `width` is more of a suggestion:
```css
.parent-container {
  background-color: #FFFCD2;
  padding: 10px;
  width: 500px;
  border: 1px solid orange;
  display: flex;
}

.child-container {
  background-color: #FF4838;
  padding: 10px;
  width: 700px;
}
```
<div style="display: flex; background-color: #FFFCD2; border: 1px solid orange; padding: 10px; width: 500px;">
    <div style="background-color: #FF4838; color: white; padding: 10px; width: 700px">Child Container</div>
</div>
<br>

`flex-basis` is another way to control the width (when `flex-direction: row`) or height (when `flex-direction: column`). Like `width`, `flex-basis` is also more of a suggestion than hard constraint:
```css
.outer-container {
  background-color: #FFFCD2;
  padding: 10px;
  color: white;
  display: flex;
  width: 500px; 
}

.left-container {
  background-color: #FF4838;
  padding: 10px;
  flex-basis: 200px;
}

.middle-container {
  background-color: #41ABFC;
  padding: 10px;
  flex-basis: 50px;  
}

.right-container {
  background-color: #1FC65D;
  padding: 10px;
  flex-basis: 50px;  
}
```
<div style="display: flex; background-color: #FFFCD2; padding: 10px; width: 500px">
    <div style="flex-basis: 200px; background-color: #FF4838; color: white; padding: 10px">Left Container</div>
    <div style="flex-basis: 50px; background-color: #41ABFC; color: white; padding: 10px">Middle Container</div>
    <div style="flex-basis: 50px; background-color: #1FC65D; color: white; padding: 10px">Right Container</div>
</div>

```css
.outer-container {
  background-color: #FFFCD2;
  padding: 10px;
  color: white;
  display: flex;
  width: 500px; 
}

.left-container {
  background-color: #FF4838;
  padding: 10px;
  flex-basis: 500px;
}

.middle-container {
  background-color: #41ABFC;
  padding: 10px;
  flex-basis: 50px;  
}

.right-container {
  background-color: #1FC65D;
  padding: 10px;
  flex-basis: 100px;  
}
```
<div style="display: flex; background-color: #FFFCD2; padding: 10px; width: 500px">
    <div style="flex-basis: 500px; background-color: #FF4838; color: white; padding: 10px">Left Container</div>
    <div style="flex-basis: 50px; background-color: #41ABFC; color: white; padding: 10px">Middle Container</div>
    <div style="flex-basis: 100px; background-color: #1FC65D; color: white; padding: 10px">Right Container</div>
</div>
<br>

`flex-grow` is another option available. It specifies how a container should expand to occupy available empty space. 0 is the default value which means the container should not grow. Setting it to one would result in the container growing to occupy all available empty space. After the containers have been placed normally, the available empty space is divided up proportionately.
```css
.left-container {
  background-color: #FF4838;
  padding: 10px;
  flex-grow: 1;
}

.middle-container {
>  background-color: #41ABFC;
  padding: 10px;
}

.right-container {
  background-color: #1FC65D;
  padding: 10px;
}
```
<div style="display: flex; background-color: #FFFCD2; padding: 10px; width: 500px">
    <div style="flex-grow: 1; background-color: #FF4838; color: white; padding: 10px">Left Container</div>
    <div style="background-color: #41ABFC; color: white; padding: 10px">Middle Container</div>
    <div style="background-color: #1FC65D; color: white; padding: 10px">Right Container</div>
</div>
<br>

Multiple containers can also set `flex-grow`. In this case, the extra space is divided proportionally between containers based on their `flex-grow` values.
```css
.left-container {
  background-color: #FF4838;
  padding: 10px;
  flex-grow: 0.5;
}

.middle-container {
  background-color: #41ABFC;
  padding: 10px;
  flex-grow: 0.25;  
}

.right-container {
  background-color: #1FC65D;
  padding: 10px;
  flex-grow: 0.25;  
}
```
<div style="display: flex; background-color: #FFFCD2; padding: 10px; width: 500px">
    <div style="flex-grow: 0.50; background-color: #FF4838; color: white; padding: 10px">1</div>
    <div style="flex-grow: 0.25; background-color: #41ABFC; color: white; padding: 10px">2</div>
    <div style="flex-grow: 0.25; background-color: #1FC65D; color: white; padding: 10px">3</div>
</div>
<br>

What happens when there isn't enough space for the child containers? They shrink proportionally:
```css
.outer-container {
  background-color: #FFFCD2;
  padding: 10px;
  color: white;
  display: flex;
  width: 300px; 
}
```
<div style="display: flex; background-color: #FFFCD2; padding: 10px; width: 300px">
    <div style="flex-grow: 0.50; background-color: #FF4838; color: white; padding: 10px">Left Container</div>
    <div style="flex-grow: 0.25; background-color: #41ABFC; color: white; padding: 10px">Middle Container</div>
    <div style="flex-grow: 0.25; background-color: #1FC65D; color: white; padding: 10px">Right Container</div>
</div>
<br>

If we don't want the container to shrink in proportion, we use `flex-shrink` property. As an example if the parent container's width is 400px, but the child containers together need 500px. There is a deficit of 100px which the child containers provide by shrinking. If all child containers have `flex-shrink: 1` they shrink in same ratio:
```css
.outer-container {
  background-color: #FFFCD2;
  padding: 10px;
  color: white;
  display: flex;
  width: 300px; 
}

.left-container {
  background-color: #FF4838;
  padding: 10px;
  flex-basis: 200px;  
  flex-shrink: 2;
}

.middle-container {
  background-color: #41ABFC;
  padding: 10px;
  flex-basis: 200px;  
  flex-shrink: 1;
}

.right-container {
  background-color: #1FC65D;
  padding: 10px;
  flex-basis: 200px;  
  flex-shrink: 1;
}
```
<div style="display: flex; background-color: #FFFCD2; padding: 10px; width: 300px">
    <div style="flex-basis: 200px; flex-shrink: 2; background-color: #FF4838; color: white; padding: 10px">1</div>
    <div style="flex-basis: 200px; flex-shrink: 1; background-color: #41ABFC; color: white; padding: 10px">2</div>
    <div style="flex-basis: 200px; flex-shrink: 1; background-color: #1FC65D; color: white; padding: 10px">3</div>
</div>
<br>

Setting `flex-shrink: 0` is often used to prevent shrinking of containers that house icons. Similar effect can also be achieved using `min-width` property:
<div style="display: flex; align-items: center; background-color: #FFFCD2; padding: 10px; width: 300px; gap: 10px">
    <div style="width: 40px; height: 40px; flex-shrink: 0; background-color: #FF4838; color: white; border-radius: 100%"></div>
    <div style="width: 300px; height: 40px; background-color: #41ABFC; color: white; padding: 10px"></div>
</div>
<br>
<div style="display: flex; align-items: center; background-color: #FFFCD2; padding: 10px; width: 200px; gap: 10px">
    <div style="width: 40px; height: 40px; flex-shrink: 0; background-color: #FF4838; color: white; border-radius: 100%"></div>
    <div style="width: 300px; height: 40px; background-color: #41ABFC; color: white; padding: 10px"></div>
</div>
<br>
<div style="display: flex; align-items: center; background-color: #FFFCD2; padding: 10px; width: 100px; gap: 10px">
    <div style="width: 40px; height: 40px; flex-shrink: 0; background-color: #FF4838; color: white; border-radius: 100%"></div>
    <div style="width: 300px; height: 40px; background-color: #41ABFC; color: white; padding: 10px"></div>
</div>
<br>

The Flexbox algorithm refuses to shrink a child below its minimum size. This means shrinking too much may lead to overflow:
```css
.outer-container {
  background-color: #FFFCD2;
  padding: 10px;
  color: white;
  display: flex;
  width: 300px; 
}

label { 
  flex-shrink: 0;
  padding: 10px;  
}

/* Input box have inherent min-width of 170px - 200px */
input {
  padding: 10px;
  border: 1px solid black;
}
```
<div style="display: flex; background-color: #FFFCD2; padding: 10px; width: 300px">
    <label for="search" style="flex-shrink: 0; padding: 10px">Search</label>
    <input type="text" id="search" style="border: 1px solid black; padding: 10px" />
</div>

## Gap
`gap` property adds space between the containers.

```css
.outer-container {
  background-color: #FFFCD2;
  padding: 10px;
  color: white;
  display: flex;
  gap: 20px;
  justify-content: center;  
}
```
