## Overview
Allows ways to select different DOM elements to which CSS styles can be applied. We can target elements in various ways - based on type, attribute, state or position in the DOM tree.

## Selectors
### Type Selector
Select using the name of tag:
```css
div {
    /* CSS styles applied to all div elements */
}

a {
    /* CSS styles applied to all anchor elements */
}
```

### Class Selector
Selects all elements that have the given `class` attribute:
```css
/*
<div class="active">...</div>
*/

.active {
    /* CSS styles applied to all elements having class active */
}
```

Can be combined with type selector:
```css
a.active {
    /* CSS styles applied to all anchor elements having class active */
}
```

### Id Selector
Selects all elements having the given `id`:
```css
/*
<p id="para-1">...</p>
*/

#para-1 {
    /* CSS styles applied to all elements having class active */
}
```

### Attribute Selector
Matches elements based on the element having a given attribute
```css
a[title] {
  /* anchor elements with a title attribute */
}

a[href="https://example.org"] {
  /* anchor elements with an href matching "https://example.org" */
}

a[href*="example"] {
  /* anchor elements with an href containing "example" */
}

a[href$=".org" i] {
  /* anchor elements with an href ending ".org", case-insensitive */
}

a[class^="container"] {
  /* anchor elements whose class attribute starts with word "container" */
}

a[class~="logo"] {
  /* anchor elements whose class attribute contains the space separated word "logo" */
}
```

## Combinators
Lets us combine selectors to select DOM elements based on their relationship to other elements within the document node tree.

### Descendent Selector
`selector1 selector2` match all `selector2` which are child of `selector1`
```css
/*
<div class="container">
    <a>
        <button>Click Me</button>
    </a>
</div>
*/

div.container button {
    /* CSS styles applied to all buttons that are descendent of div element */
}
```

### Child Selector
More specific than descendent selector, selects immediate child element:
```css
a > button {
    /* CSS styles applied to all buttons that are direct child of anchor element */
}
```

### Sibling Selector
The *subsequent sibling* selector selects sibling elements:
```css
/*
<div class="container">
    <h1>Header</h1>
    <p>Some text here</p>
    <span> Some more text</span>
</div>
*/

h1 ~ span {
    /* CSS styles applied to all spans that are sibling of h1 element appearing later */
}
```

Next sibling selector is more specific and selects immediate adjacent sibling element:
```css
h1 + p {
     /* CSS styles applied to paragraph element that is immediate next sibling of h1 element */
}
```

## Pseudo Classes
Keyword added to a selector that lets us select elements based on information that lies outside of the document tree (like current state of the element).

### Input Pseudo Classes
- `:enabled`, `:disabled`: user interface element that is in an enabled or disabled state

  ```css
  /*
      <input disabled type="text" />
  */
  
  input:disabled {
      /* CSS styles applied to all input elements that are disabled */
  }
  ```
- `:checked`: matches when elements such as checkboxes and radio buttons are toggled on.

  ```css
  <input type="checkbox" />

  input:checked {
      /* CSS styles applied to all checkbox elements that are checked */
  }
  ```
- `:required`: matches when a form element is required.

### Location Pseudo Classes
- `:link`: matches links that have not yet been visited.
- `:visited`: matches links that have been visited.

### Tree-structural Pseudo Classes
- `:empty`: represents an element with no children other than white-space characters.
- `:nth-child()`: notation to select elements from a list of sibling elements.

  ```css
  /*
      <ul>
        <li>One</li>
        <li>Two</li>   <!-- selected -->
        <li>Three</li>
      </ul>
  */
  
  li:nth-child(2) {
    /* CSS styles applied to 2nd child li element */
  }
  ```
- `:first-child`: matches an element that is the first of its siblings.
- `:last-child`: matches an element that is the last of its siblings.

### User Action Pseudo Classes
- `:hover`: matches when a user designates an item with a pointing device, such as holding the mouse pointer over the item.
- `:active`: matches when an item is being activated by the user. For example, when the item is clicked on.
- `:focus`: matches when an element has focus.

## Pseudo Elements
Keyword added to a selector that lets you style a specific part of the selected element
- `::first-line`: the first line-box of the originating element.
- `::first-letter`: the first letter, number, or symbol character on the first line of its originating element.
- `::selection`: applies to portion of document that has been selected
- `::before`: creates a pseudo-element that is the first child of the selected element.
- `::after`: creates a pseudo-element that is the last child of the selected element.

  ```css
    /*
  before and after let us insert content before or after an element’s actual content—without changing the HTML.
  */
  
  blockquote::before {
    content: "“";
  }
  
  blockquote::after {
    content: "”";
  }
  ```
