## Box Model
The four aspects that make up the box model are:
1. Content
2. Padding
3. Border
4. Margin

When we say that an element should have a `width` of 100%, what does that actually mean? Check the example below:
```html
<style>
  section {
    width: 500px;
  }
  .box {
    width: 100%;
    padding: 20px;
    border: 4px solid;
  }
</style>
<section>
  <div class="box"></div>
</section>
```

When we set our `.box` to have `width: 100%`, we're saying that the box's content size should be equal to the available space, 500px. The padding and border is added on top. The box winds up being 548px wide because it adds 20px of padding and 4px of border to each side: 500 + 20 * 2 + 4 * 2.

The `box-sizing` CSS property allows us to change the rules for size calculations. The default value (`content-box`) only takes the inner content into account, but it offers an alternative value: `border-box`. With `box-sizing: border-box`, things behave much more intuitively. We can set this as default by:
```css
*,
*::before,
*::after {
  box-sizing: border-box;
}
```

## Padding
Padding is "inner space". Padding can be set for all directions at once, or it can be specified for individual directions:
```css
.even-padding {
  padding: 20px;
}
.asymmetric-padding {
  padding-top: 20px;
  padding-bottom: 40px;
  padding-left: 60px;
  padding-right: 80px;
}
/* The same thing, but using ✨ logical properties ✨ */
.asymmetric-logical-padding {
  padding-block-start: 20px;
  padding-block-end: 40px;
  padding-inline-start: 60px;
  padding-inline-end: 80px;
}
```

Just like fonts, we can use `px`, `em` or `rem` as units, however it is better to use pixels as units for padding, margin and borders as opposed to em/rem for texts. We can also use percentage, however the result is surprising and counter-intuitive.

There are shorthand ways of specifying padding:
```css
/* Top/bottom then left/right padding */
.two-way-padding {
  padding: 15px 30px;
}

/* Top, right, bottom, left */
.asymmetric-padding {
  padding: 10px 20px 30px 40px;
}
```

## Border
Border is a bit of an odd duck in the trinity of padding/border/margin—unlike the other two, it has a visual/cosmetic components:
- Border width (eg. `3px`, `1em`)
- Border style (eg. `solid`, `dotted`)
- Border color (eg. `red`, `blue`)

Shorthand version is:
```css
/* Only required component is the border-style */
.box {
  border: 3px solid yellow;
}

/* Won't work – needs a style! */
.not-good {
  border: 2px pink;
}
```

If we don't specify a border color, it'll use the font's color by default. This isn't well-known, but it can be useful in cases where those things should be synchronized. If we want to specify this behaviour explicitly, it can be done with the special `currentColor` keyword. `currentColor` is always a reference to the element's derived text color (whether set explicitly or inherited), and it can be used anywhere a color might be used:

The `border-radus` property should have been called `corner-radius`. To specify border radius:
```css
.box {
	border-radius: 25px;
}

/* Top-left, top-right, bottom-right, bottom-left */
.box-another {
	border-radius: 10px 10px 20px 20px;
}

/* Also accept percentage, 50% creates circle */
.circle {
	border-radius: 50%;
}
```

**Border vs Outline:** both add a visual edge to a given element. The core difference is that outline doesn't affect layout. Outline is kinda more like box-shadow; it's a cosmetic effect draped over an element, without nudging it around, or changing its size. Outlines are stacked outside border, and can sometimes be used as a "second border", for effect:
```css
.box {
  width: 100px;
  height: 100px;
  border: 4px solid darkviolet;
  outline: 4px solid deeppink;
}
```
<div style="width: 100px; height: 100px; border: 4px solid darkviolet; outline: 4px solid deeppink;"></div>
<br>

Outlines have a special `outline-offset` property. It allows you to add a bit of a gap between the element and its outline.
```css
.box {
  width: 100px;
  height: 100px;
  border: 4px solid darkviolet;
  outline: 4px solid currentColor;
  outline-offset: 4px;
}
```
<div style="width: 100px; height: 100px; border: 4px solid darkviolet; outline: 4px solid deeppink; outline-offset: 4px;"></div>
<br>

It is important that we don’t set `outline: none`. The reason being that this totally breaks navigation for keyboard users. That ring is required for them to know which element is currently focused! The only exception is if we provide a suitable alternative. For example:
```css
button {
  outline: none;
}

button:focus {
  background: navy;
  color: white;
}
```

## Margin
Margin increases the space around an element, giving it some breathing room. The syntax for `margin` is same as that for `padding`. One interesting aspect about margin is that it can accept negative values (not allowed for padding or border but allowed for outline-offset). A negative margin can pull an element outside its parent.
Margin is about changing the gap between elements. Negative margin shrinks the gap below an element, causing the next element to scoot up closer. Consider the example:

```css
.outer-container {
  width: 250px; 
  padding: 10px; 
  border: 1px solid black;
}

p {
  border: 1px solid black;  
}
```
<div style="width: 250px; padding: 10px; border: 1px solid black">
    <p style="border: 1px solid black">First paragraph with bit of text in it to provide some content.</p>
    <p style="border: 1px solid black">Second paragraph with bit of text in it to provide some content.</p>
    <p style="border: 1px solid black">Third paragraph with bit of text in it to provide some content.</p>
</div>

To start, let’s give the first paragraph a -5px `margin-bottom`. Essentially, when the browser calculates the point where the second paragraph should start, it moves that point 5px upward. From then on the browser lays out all paragraphs as normal. Therefore the second paragraph, being the bottom neighbour of the first one, is drawn 5px closer to the first paragraph. The margin between the second and third paragraphs remains intact:
```css
.outer-container {
  width: 250px; 
  padding: 10px; 
  border: 1px solid black;
}

p {
  border: 1px solid black;  
}

p.first {
  margin-bottom: -5px;
}
```
<div style="width: 250px; padding: 10px; border: 1px solid black">
    <p style="border: 1px solid black; margin-bottom: -5px;">First paragraph with bit of text in it to provide some content.</p>
    <p style="border: 1px solid black">Second paragraph with bit of text in it to provide some content.</p>
    <p style="border: 1px solid black">Third paragraph with bit of text in it to provide some content.</p>
</div>

f we had given the second paragraph `margin-top: -5px` the effect would have been the same.

*Margin collapsing* behaves different when you use negative margins:
> In the case of negative margins, the maximum of the absolute values of the negative adjoining margins is deducted from the maximum of the positive adjoining margins. If there are no positive margins, the maximum of the absolute values of the adjoining margins is deducted from zero.

Normally, the browser would take the first paragraph’s margin-bottom and the second one’s margin-top, figure out which one is larger, and apply that margin between the two, which would yield max(-15px,1em) = 1em. That’s not how it works, though. In case of negative margins we take the absolute value of the negative margin (15px) and deduct it from the positive margin (1em). This yields about 1px (depending on the font size, of course).

Negative `margin-left` and `margin-right` work the same, provided the element **has a width**. Here we apply `margin-left: -10px` and `margin-right: 10px`:
```css
.outer-container {
  width: 250px; 
  padding: 10px; 
  border: 1px solid black;
}

p {
  border: 1px solid black;  
}

p.first {
  margin-left: -10px;
}

p.second {
  margin-right: -10px;
}

/* Add both margins to stretch the container */
p.third {
  margin-left: -10px;
  margin-right: -10px;
}
```
<div style="width: 250px; padding: 10px; border: 1px solid black">
    <p style="border: 1px solid black; margin-left: -10px;">First paragraph with bit of text in it to provide some content.</p>
    <p style="border: 1px solid black; margin-right: -10px;">Second paragraph with bit of text in it to provide some content.</p>
    <p style="border: 1px solid black; margin-left: -10px; margin-right: -10px;">Third paragraph with bit of text in it to provide some content.</p>
</div>
<br>

**Margin Collapse:** some rules regarding margin collapse:
1. Only vertical margins collapse:
   ```css
   div {
     border: 1px solid black;
     padding-left: 10px;
     padding-right: 10px;
     width: 300px;
   }
   
   p {
     border: 1px solid black;
     margin-top: 10px;
     margin-bottom: 10px;
   }
   ```
   <div style="border: 1px solid black; margin-top: 10px; padding-right: 10px; padding-left: 10px; width: 300px;">
       <p style="border: 1px solid black; margin-top: 10px; margin-bottom: 10px;">Paragraph One</p>
       <p style="border: 1px solid black; margin-top: 10px; margin-bottom: 10px;">Paragraph Two</p>
   </div>
   <br>

3. Margin collapse is unique to *Flow* layout. If we have children inside a `display: flex` parent, those children's margins will never collapse.

4. Only adjacent elements collapse:
   ```css
   div {
     border: 1px solid black;
     padding-left: 10px;
     padding-right: 10px;
     width: 300px;
   }
   
   p {
     border: 1px solid black;
     margin-top: 10px;
     margin-bottom: 10px;
   }
   ```
   <div style="border: 1px solid black; margin-top: 10px; padding-right: 10px; padding-left: 10px; width: 300px;">
       <p style="border: 1px solid black; margin-top: 10px; margin-bottom: 10px;">Paragraph One</p>
       <hr />
       <p style="border: 1px solid black; margin-top: 10px; margin-bottom: 10px;">Paragraph Two</p>
   </div>
   <br>

5. The bigger margin wins

6. Nesting doesn't prevent collapsing: margin will always try and increase distance between siblings, **even if it means *transferring* margin to the parent element!** In the below case, the effect is the same as if we had applied the margin to the parent `<div>`, not the child `<p>`.
   ```html
   <style>
      div {
        border: 1px solid black;
        padding-left: 10px;
        padding-right: 10px;
        width: 300px;
      }
       
      p {
        border: 1px solid black;  
        margin-top: 10px;
        margin-bottom: 10px;
      }
    </style>
    <div class="outer">
      <div>
        <p>Paragraph One</p>
      </div>
      <p>Paragraph Two</p>
    </div>
   ```
   <div style="border: 1px solid black; margin-top: 10px; padding-right: 10px; padding-left: 10px; width: 300px;">
       <div>
           <p style="border: 1px solid black; margin-top: 10px; margin-bottom: 10px;">Paragraph One</p>
       </div>
       <p style="border: 1px solid black; margin-top: 10px; margin-bottom: 10px;">Paragraph Two</p>
   </div>
   <br>

7. Blocked by padding or border: we can think of padding/border as a sort of wall; if it sits between two margins, they can't collapse, because there's an obstruction in the way.

8. Margins can collapse in the same direction
   ```html
   <style>
     .parent {
       margin-top: 72px;
     }
     .child {
       margin-top: 24px;
     }
   </style>
   <div class="parent">
     <p class="child">Paragraph One</p>
   </div>
   ```

**Auto Margins:** can be used to center a child in a container. Example:
```html
<main>
  <section class="content">
    Hello World
  </section>
</main>

<style>
.content {
  width: 50%;
  margin-left: auto;
  margin-right: auto;
  background: palevioletred;
  padding: 16px;
}

main {
  width: 500px;
  height: 70px;
  border: 3px solid silver;
  background: yellow;
}
</style>
```
<div style="width: 500px; height: 70px; border: 3px solid silver; background: yellow;">
  <section style="width: 50%; margin-left: auto; margin-right: auto; background: palevioletred; padding: 16px;">
    Hello World
  </section>
</div>

The `auto` value seeks to fill the *maximum available space* (similar to `width` property). When we set both `margin-left` and `margin-right` to `auto`, we're telling them each to take up as much space as possible.
Two caveats:
- This only works for horizontal margin. Setting top/bottom margin to `auto` is equivalent to setting it to `0px`.
- This only works on elements with an explicit `width`. Block elements will naturally grow to fill the available horizontal space, so we need to give our element a `width` in order to center it.
