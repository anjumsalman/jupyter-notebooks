## Overview
In CSS a variable is any property whose name begins with `--`. As an example:
```css
p {
    /* variable name is case sensitive */
    --color-yellow: yellow;
}
```

In the above example, the variable `--color-yellow` is scoped to the `p` selector. However, we can also define global variable by using the `:root` pseudo-class:
```css
:root {
  --primary-text-color: red
}
```

To use variable we make use of `var()` function:
```css
a {
    color: var(--primary-text-color);
}
```

CSS variables are resolved with the normal inheritance and cascade rules:
```css
div {
  --color: red;
}

/* --color available in any div */

div.test {
  color: var(--color)
}

div.ew {
  color: var(--color)
}
```

The values also react to media queries:
```css
:root {
 --gutter: 10px 
}

@media screen and (min-width: 768px) {
    --gutter: 30px
}
```
