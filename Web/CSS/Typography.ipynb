## Font Family
`font-family` property allows us to set a font for textual content. It's called a *family* because each font consists of multiple character sets: for example, “Roboto” includes 12 individual sets: 6 font weights, with 2 variants (normal and italic). Font families come in different styles. The 3 most popular are serif, sans-serif and monospace. If we'd like, we can let the operating system pick a reasonable default for each style:
```css
p {
  font-family: sans-serif;
}
```

We can pass multiples values (a *font stack*) so that the browser can fall back to an available font for the first few seconds (or if the font fails to download).
```css
article {
	font-family: 'Roboto', sans-serif;
}
```

## Formatting
**Bold**, *italics* and <ins>underline</ins> are the three most commonly used formatting options. For bold we can either use textual value or a numeric value (1-1000). Default value is 400 and for bold, it is 700.
```css
p {
	font-weight: 700 /* equivalent to font-weight:bold; */
}
```

If we only supply a single font weight (in the font family), the browser will do its best to represent bold text by thickening the characters. It generally doesn't do a great job at this. Similar to bold text, the browser can simulate italic text by rendering the characters at an angle. For best results, though, we should supply an italic character set.

To add strikethrough or underline, we use `text-decoration` property:
```css
.text {
  text-decoration: underline;         /* underline */
  /* text-decoration: none;  to remove (e.g., on links) */
  text-decoration: line-through;      /* strikethrough */
  text-decoration: overline;          /* line above */
}
```

The long form lets us change the line style as well:
```css
text-decoration: underline 2px red;
```

For italics, we use the `font-style` property:
```css
.text {
  font-style: italic;    /* italicize */
  font-style: normal;    /* reset (also "oblique" — slanted without trueics) */
}
```

## Transformation
To change the case of our text, we can use text-transform property:
```css
/* RENDER WITH ALL CAPS */
text-transform: uppercase;
/* Capitalize The First Letter Of Every Word */
text-transform: capitalize;
```

## Alignment
We can shift characters horizontally using the `text-align` property. It can take values like left, right and center. `text-align` is also capable of aligning other elements, like images. In general, though, we'll use other tools for those kinds of jobs. We should reserve `text-align` for text.

## Spacing
We can tweak the spacing of our characters in two ways.
1. We can tweak the horizontal gap between characters using the `letter-spacing` property.
2. We can tweak the vertical distance between lines using the `line-height` property.

```css
p {
  letter-spacing: 15px
}
```

    <p style="letter-spacing: 15px">Hello</p>
</div>
<br>

`line-height` controls the height of a line box which is the vertical space each line of text occupies. It's not the font size itself, but the total space "reserved" for a line, and it determines how tightly or loosely text is spaced. Its value is specified as a number (or pixels or percentage). When the unitless number is used the line height becomes multiple of that number. Example: if font size is 16px and line height is 1.5, then the value calculated is 24px.  

<img src="./images/line-height.png" width=500 height=auto>

Originally, the default value for this property was `1`, but it's ticked up over the years: in Chrome, the new default value is `1.15`. In Firefox, it's `1.2`. Both of these values are too low. In order to make our apps and websites as accessible as possible, we want to choose a pretty generous value for `line-height`. Many places recommend its minimum value as 1.5.
