## Tools
AI Agents can use variety of tools to accomplish its goalds like:
- searching the web
- calling APIs
- doing filesystem operations, etc

As an example, though LLMs can answer math questions, it may or may not be correct. Thus it is better that it uses a tool that performs mathematic calculations which the LLM can then use.

Use `@Tool` to annotate a method that provides tool functionality:

```java
@Tool("Returns a square root of a given number")
double squareRoot(double x) {
    return Math.sqrt(x);
}
```

To increase the chances of the LLM calling the right tool with the right arguments, we should provide a clear and unambiguous:
- name of the tool
- description of what the tool does and when it should be used
- description of every tool parameter

Example:

```java
class WeatherTool {
    @Tool("""
          Retrieves the current weather forecast for a specific city and date.
          Use this whenever the user asks about weather conditions, temperature,
          or whether they need an umbrella/jacket for a given place and day.""")
    String getWeather(
            @P("The city name, e.g. 'Paris' or 'San Francisco'. Do not include country or state.")
            String city,

            @P("""
               The date to check, in ISO-8601 format (yyyy-MM-dd).
               Use today's date if the user doesn't specify one.""")
            String date
    ) {
        return weatherService.fetch(city, date);
    }
}
```

**Optional Parameter:** By default all the parameters are considered as required, to mark a parameter as optional:

```java
@Tool
String getTemperature(String location, @P(required = false) Unit unit) {
    ...
}

@Tool
String getTemperature(String location, Optional<Unit> unit) {
    ...
}
```

To make a property of complex type as optional:

```java
// @JsonProperty marks email as optional
record User(String name, @JsonProperty(required = false) String email) {}

@Tool
void add(User user) {
    // ...
}
```

**Default Parameter:**

```java
@Tool
List<Article> searchArticles(
    String query,
    @P(defaultValue = "10") int limit,
    @P(defaultValue = "[\"en\"]") List<String> languages
) {
    // ...
}
```
